```python
# TG-AudioSelector Telegram Bot in a Single Google Colab Cell (Fixed Version)
# Repository: https://github.com/abhinai2244/TG-AudioSelector
# Fixed: Syntax error in original config.py (trailing comma in ALLOWED_GROUP_IDS).
#        Now sets ALLOWED_GROUP_IDS = [] by default (add group IDs manually if needed).
#        Updated replacements to match actual config.py format.
# Instructions:
# 1. Run this cell. It will prompt for your Telegram API credentials.
# 2. Ensure you have API_ID, API_HASH from https://my.telegram.org and BOT_TOKEN from @BotFather.
# 3. The bot will start running at the end. Interact via Telegram.
# Notes: Colab is temporary; for production, use a VPS. FFmpeg is installed for audio processing.
#        If you need to allow specific groups, edit config.py after setup and add IDs to ALLOWED_GROUP_IDS.

import os
import subprocess
import sys
import re

# Step 1: Install system dependencies (FFmpeg)
subprocess.run(['apt', 'update', '-qq'], capture_output=True)
subprocess.run(['apt', 'install', '-qq', 'ffmpeg'], capture_output=True)
print("FFmpeg installed.")

# Step 2: Clone repository
subprocess.run(['git', 'clone', 'https://github.com/abhinai2244/TG-AudioSelector.git'], capture_output=True)
os.chdir('TG-AudioSelector')
print("Repository cloned. Current dir:", os.getcwd())
subprocess.run(['ls', '-la'], shell=True, capture_output=True)

# Step 3: Install Python dependencies
subprocess.run([sys.executable, '-m', 'pip', 'install', '-U', '-r', 'requirements.txt'], capture_output=True)
print("Python dependencies installed: pyrogram, tgcrypto, etc.")

# Step 4: Interactive configuration of config.py
print("\n--- CONFIGURATION ---")
print("Edit config.py with your credentials. We'll do it interactively.")
print("Note: Fixed syntax error in ALLOWED_GROUP_IDS (set to empty list; add group IDs if needed).")

# Backup original config
subprocess.run(['cp', 'config.py', 'config.py.backup'])

# Read original content
with open('config.py', 'r') as f:
    content = f.read()

# Fix the syntax error in ALLOWED_GROUP_IDS first
# Replace the invalid list with a valid empty list
bad_list_pattern = r'ALLOWED_GROUP_IDS = \[\s*,\s*# Your group ID from logs\s*\n\s*# Add more group IDs as needed\s*\]'
content = re.sub(bad_list_pattern, 'ALLOWED_GROUP_IDS = []  # Add your group IDs from logs', content)

# Get user inputs
api_id = input("Enter your API_ID: ").strip()
api_hash = input("Enter your API_HASH: ").strip()
bot_token = input("Enter your BOT_TOKEN: ").strip()
owner_id = input("Enter your OWNER_ID (your Telegram user ID): ").strip()

# Replace placeholders (matched to actual format)
content = content.replace("API_ID =   # Replace with your API ID", f"API_ID = {api_id}")
content = content.replace('API_HASH = ""  # Replace with your API Hash', f'API_HASH = "{api_hash}"')
content = content.replace('BOT_TOKEN = ""  # Replace with your Bot Token', f'BOT_TOKEN = "{bot_token}"')
content = content.replace("OWNER_ID =   # Owner's user ID", f"OWNER_ID = {owner_id}")

# Optional: Update PREMIUM_USERS (original is {5756495153}, so append or replace)
premium_input = input("Add yourself to PREMIUM_USERS? (y/n): ").strip().lower()
if premium_input == 'y':
    # Append to existing set (simple string replace to add)
    content = re.sub(r'PREMIUM_USERS = \{5756495153\}', f'PREMIUM_USERS = {{5756495153, {owner_id}}}', content)
    print(f"Added {owner_id} to PREMIUM_USERS.")

# Write back
with open('config.py', 'w') as f:
    f.write(content)

print("config.py updated and fixed! Preview first 500 chars:")
preview = open('config.py', 'r').read()[:500]
print(preview + "..." if len(preview) > 500 else preview)
print("\nIf incorrect, edit manually via file browser or !cat config.py")

# Step 5: Run the bot
print("\n--- STARTING BOT ---")
print("Bot is now running. Use /start in Telegram. Interrupt kernel to stop.")
print("Logs below...")

# Run main.py using subprocess and stream output
process = subprocess.Popen([sys.executable, 'main.py'], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, universal_newlines=True, bufsize=1)

# Stream output in real-time
while True:
    output = process.stdout.readline()
    if output == '' and process.poll() is not None:
        break
    if output:
        print(output.strip())

# On exit
rc = process.poll()
if rc == 0:
    print("Bot exited normally.")
else:
    print(f"Bot exited with code {rc}. Check logs above for errors.")
```
